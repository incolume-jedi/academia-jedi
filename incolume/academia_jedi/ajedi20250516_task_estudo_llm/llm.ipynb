````python
from config import settings
import os
import langchain
````

````python

from langchain_openai import OpenAI
langchain.verbose = False
langchain.debug = False
langchain.llm_cache = False
````

É necessário configurar o token de acesso nas variáveis de ambiente ou arquivos de segredos.

Neste projeto é configurado a biblioteca `dynaconf` para gerenciar variáveis de ambiente e segredos.

Na raiz do projeto, no arquivo `settings/secrets.yml` acrescente o token, conforme abaixo:

``` yalm
default:
  OPENAI_API_KEY: value-token-openai-api-key
``` 

````python

os.environ['OPENAI_API_KEY'] = settings.openai_api_key
````

````python
llm = OpenAI()
````

````python
pergunta = 'Conte a história do linux em até 20 linhas.'
llm.invoke(pergunta)
````

````python
for trecho in llm.stream(pergunta):
    print(trecho, end='')
````
