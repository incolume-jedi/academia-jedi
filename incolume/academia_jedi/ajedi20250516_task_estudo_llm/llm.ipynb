## PromptTemplates uso 

````python
from langchain.prompts import PromptTemplate

# Criando um PromptTemplate
prompt_template = PromptTemplate.from_template('Responda a seguinte pergunta: {pergunta}')

# Usando o PromptTemplate
resultado = prompt_template.format(pergunta='O que é um buraco negro?')
print(resultado)
````

````python
# Usando o PromptTemplate
resultado = prompt_template.invoke({'pergunta': 'O que é um buraco negro?'})
print(resultado)
````

````python
from langchain.prompts import ChatPromptTemplate

# Criando um ChatPromptTemplate
chat_template = ChatPromptTemplate.from_template('Essa é a minha dúvida: {duvida}')

# Formatando uma mensagem
mensagens = chat_template.format_messages(duvida='Quem sou eu?')
print(mensagens)
````

````python
from langchain_core.messages import HumanMessage, AIMessage, SystemMessage
from langchain.prompts import ChatPromptTemplate

chat_template = ChatPromptTemplate.from_messages(
    [SystemMessage(content='Você é um assistente útil.'),
    HumanMessage(content='Qual é a capital da França?'),
    AIMessage(content='A capital da França é Paris.')]
)

display(chat_template)
````

````python
chat_template = ChatPromptTemplate.from_messages(
    [('system', 'Você é um assistente útil.'),
     ('human', 'Qual é a capital da França?'),
     ('ai', 'A capital da França é Paris.')]
)

display(chat_template)
````

# HuggingFace

````python
from langchain_community.chat_models.huggingface import ChatHuggingFace
from langchain_community.llms.huggingface_endpoint import HuggingFaceEndpoint

modelo = 'mistralai/Mixtral-8x7B-Instruct-v0.1'
llm = HuggingFaceEndpoint(repo_id=modelo)
chat = ChatHuggingFace(llm=llm)

mensagens = [
    HumanMessage(content='Quanto é 1 + 1?'),
]
resposta = chat.invoke(mensagens)

resposta
````

É necessário configurar o token de acesso nas variáveis de ambiente ou arquivos de segredos.

Neste projeto é configurado a biblioteca `dynaconf` para gerenciar variáveis de ambiente e segredos.

Na raiz do projeto, no arquivo `settings/secrets.yml` acrescente o token, conforme abaixo:

