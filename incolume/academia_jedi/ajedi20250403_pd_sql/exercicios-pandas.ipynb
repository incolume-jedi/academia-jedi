### 12. Você conseguiria obter uma tabela contendo a variação percentual ano a ano para o estado do Rio de Janeiro?

```python
rj = df['ESTADO'] == 'RIO DE JANEIRO'
media_anual_rj = (
    df[rj & gc]
    .groupby(by=[df['DATA FINAL'].dt.year])['PREÇO MÉDIO REVENDA']
    .mean()
)
media_anual_rj
```

```python
"""Ultimo preço do ano."""
rj_ano_gc = (
    df[rj & gc]
    .groupby(by=[df['DATA FINAL'].dt.year])['PREÇO MÉDIO REVENDA']
    .last()
)
rj_ano_gc
```

### Deslocamento de valores

```python
rj_ano_gc.shift(1)
```

```python
rj_indice_aumento_anual = rj_ano_gc / rj_ano_gc.shift(1) - 1
rj_indice_aumento_anual = rj_indice_aumento_anual.fillna(0).apply(
    lambda x: f'{x:.2%}',
)
rj_indice_aumento_anual
```

### DESAFIO: Crie uma tabela contendo uma serie temporal com a diferença absoluta e percentual entre os valores mais baratos e caros. Apresente também ao lado os estados na qual os maiores e menores preços foram registrados.

```python
df[gas].groupby(by='AAAA-MM').max()['PREÇO MÉDIO REVENDA']
df[gas].groupby(by='AAAA-MM').min()['PREÇO MÉDIO REVENDA']
desafio = pd.DataFrame(
    data={
        'MAX': df[gas].groupby(by='AAAA-MM').max()['PREÇO MÉDIO REVENDA'],
        'MIN': df[gas].groupby(by='AAAA-MM').min()['PREÇO MÉDIO REVENDA'],
    },
)
desafio['PREÇO_MÉDIO'] = desafio.MAX / desafio.MIN
desafio['ABS_DIFF'] = desafio.MAX - desafio.MIN
desafio['PERCENT_DIFF'] = desafio.ABS_DIFF / desafio.MIN
idx_max = df.groupby('AAAA-MM')['PREÇO MÉDIO REVENDA'].idxmax()
idx_min = df.groupby('AAAA-MM')['PREÇO MÉDIO REVENDA'].idxmin()
display(df.loc[idx_max, 'ESTADO'])
desafio['MIN_VALUE_UF'] = df.loc[idx_min, 'ESTADO'].values
desafio['MAX_VALUE_UF'] = df.loc[idx_max, 'ESTADO'].values
desafio
```

### Quais estado possuem maiores preços?

```python
desafio.MAX_VALUE_UF.value_counts()
```

### Quais estados com menores preços

```python
desafio.MIN_VALUE_UF.value_counts()
```
