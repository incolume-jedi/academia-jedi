## Scatter plots

```python
import plotly.graph_objects as go
import numpy as np
```

```python
t = np.linspace(0, 10,100)
y = np.sin(t)
```

```python
fig = go.Figure(data=go.Scatter(x=t, y=y, mode='markers'))
fig.show()
```

```python
fig = go.Figure(data=go.Scatter(x=t, y=y, mode='lines'))
fig.show()
```

```python
fig = go.Figure(data=go.Scatter(x=t, y=y, mode='lines+markers'))
fig.show()
```

```python
N=100
x = np.linspace(0, 1, N)
random_y0 = np.random.randn(N) + 5
random_y1 = np.random.randn(N)
random_y2 = np.random.randn(N) -5
```

```python
fig = go.Figure()
fig.add_trace(go.Scatter(x=x, y=random_y0, mode='markers', name='marker'))
fig.add_trace(go.Scatter(x=x, y=random_y1, mode='lines+markers', name='lines+marker'))
fig.add_trace(go.Scatter(x=x, y=random_y2, mode='lines', name='lines'))
```

## Grafico de bolha

```python
import plotly.graph_objects as go

fig = go.Figure(
    data=go.Scatter(
        x=[1,2,3,4],
        y=[10,11,12,13],
        mode='markers',
        marker=dict(size=[40, 60, 80, 100], color=[1,2,3,4])
    )
)

fig.show()
```

```python
fig = go.Figure(
    data=go.Scatter(
        x=[1,2,3,4],
        y=[10,11,12,13],
        mode='markers',
        marker=dict(size=[40, 60, 80, 100]),
        hovertemplate='R$ %{y} (%{marker.size})'
    )
)

fig.show()
```

# Bar Charts

```python
import plotly.graph_objects as go
```

```python
animais=['Girafas', 'Macados', 'Tigres']
```

```python
fig = go.Figure(
    data= [
        go.Bar(x=animais, y=[3, 120, 6], name='Zoo DF'),
        go.Bar(x=animais, y=[1, 40, 2], name='Zoo RS'),
    ]
)
fig.update_layout(barmode='stack')
fig.show()
```

```python
cores = ['lightslategray'] * 5
cores[1] = 'blue'
fig = go.Figure(
    data=[go.Bar(
        x=['Item A', 'Item B', 'Item C', 'Item D', 'Item E'],
        y=[20,14,23,25,22],
        marker_color= cores,
    )]
)

fig.show()
```

## Pie Charts

```python
labels = ['Oxigênio', 'Hidrogênio', 'Carbono', 'Nitrogênio']
values = [1053, 4500, 2500, 500]

fig = go.Figure(
    data=go.Pie(labels=labels, values=values)
)

fig.show()
```

```python
fig = go.Figure(
    data=go.Pie(labels=labels, values=values)
)
fig.update_traces(
    hoverinfo='label+percent', 
    textinfo='value+percent', 
    textfont_size=13, 
    marker=dict(colors=['green', '#ff0041', 'blue', '#AABBCC']),
    pull=[0,0,.2,0]
    )

fig.show()
```

## Candlesticks

```python
import plotly.graph_objects as go
import pandas as pd
```

```python
from pathlib import Path

finance_file = Path.cwd().parents[2].joinpath('data_files', 'csv', 'finance-chats-apple.csv')
finance_file.is_file()
```

```python
try:
    df0 = pd.read_csv('https://raw.githubusercontent.com/plotly/datasets/master/finance-charts-apple.csv')
except Exception:
    df0 = pd.read_csv(finance_file)
df0
```

```python
fig = go.Figure(
    data=go.Candlestick(
        x=df0['Date'],
        open=df0['AAPL.Open'],
        high=df0['AAPL.High'],
        low=df0['AAPL.Low'],
        close=df0['AAPL.Close'],
    )
)

fig.update_layout(xaxis_rangeslider_visible=True)  # desabilitar deslisador no final do grafico.

fig.show()
```
